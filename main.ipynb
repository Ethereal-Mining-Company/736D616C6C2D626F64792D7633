## Problem Statement & Hypothesis Formulation

### I. [Problem Statement](main.ipynb#problem-statement)
<a id="problem-statement"></a>
*State the occurrence that you are attempting to describe using data science* \
Two ways to predict the `diameter` of a small-body are **(1)** using a combination of `very high data collection cost`, `very high predictive power`, and `low observational reliability` features like `albedo` (Group 3) with `low data collection cost`, `medium observational reliability` features, like `absolute magnitude` (`H`) (Group 2) and **(2)** using a combination of `low data collection cost`, `high observational reliability` features (Group 1) with `low data collection cost`, `medium observational reliability` features, like `absolute magnitude` (`H`) (Group 2)(Ge, 2026; Murcia, 2023; Robinson, 2024). The project looks to determine if the `diameter` of a small-body can accurately be predicted with only Group 1 features to maintain the low-cost, high-reliability of features that are inexpensive and not variable or prone to error while observing.

### II. [Hypothesis Formulation](main.ipynb#hypothesis-formulation)
*Suggested explanation to the problem statement.* \
The project utilizes a two-model comparison framework that will be assessed from a statistical and operational perspective. The regression models `target` `diameter` (`Y`) using two feature subsets: an expanded set of `low-cost`, `high-reliability` observations with `low-cost`, `medium-reliability` observations (**X**<sub>full</sub>) and a restricted set of only `low-cost`, `high reliability` observations (**X**<sub>restricted</sub>). The restricted model (**X**<sub>restricted</sub>) is preferred due to the lower number and higher realiability of features. Current peer-reviewed research dictates that `absolute magnitude` (`H`), a Group 2 feature, is mathematically required to calculate `diameter` (Ge, 2026). 

The hypothesis determines if small-body `diameter` (`Y`) can be accurately predicted within a statistical and operational threshold ($\delta$) by the restrictive feature subset compared to the full feature subset, quantified as:
<div align="center">
  &Delta; R<sup>2</sup><sub>test</sub> = R<sup>2</sup><sub>Model(<b>X</b><sub>full</sub>)</sub> - R<sup>2</sup><sub>Model(<b>X</b><sub>restricted</sub>)</sub>
</div>
<br>

**Statistical Hypothesis** \
The statistical hypothesis tests if the restrictive feature subset of Group 1 features maintains a statistically acceptable degree of accuracy in predicting `target` `diameter` (`Y`) relative to the expanded feature subset of Group 1 and Group 2 features.
* **Statistical Null Hypothesis (H<sub>0,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &le; 0 <br> The inclusion of Group 2 features yield no improvement or a decrease in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do not** increase the accuracy of predicting the `target` `diameter` (`Y`).
* **Statistical Alternative Hypothesis (H<sub>1,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; 0 <br> The inclusion of Group 2 features yields an increase in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do** increase the accuracy of predicting the `target` `diameter` (`Y`).

**Operational Hypothesis** \
The operational hypothesis tests if the restrictive feature subset of Group 1 features maintains an operationally and practically acceptable degree of accuracy ($\delta$) in predicting `target` `diameter` (`Y`) relative to the expanded feature subset of Group 1 and Group 2 features.
* **Operational Null Hypothesis (H<sub>0,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; &#948; <br> The inclusion of Group 2 features yield an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is not** sufficient enough for operational use.
* **Operational Alternative Hypothesis (H<sub>1,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &le; &#948;<br> The inclusion of Group 2 features does not yields an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is** sufficient enough for operational use.

### II. Prepare
<a id="prepare"></a>
*Explore and pre-process.*
Data has been prepared for regression analysis in the following ways:
1. **Data Structure Integrity:** Validate all data types are correct. Name columns should be strings (`str`), continuous values should be floats (`float64`), and discrete values should be integers (`int`).
1. **Data Domain Validation:** Remove data is implausible such as negative `diameter`.
1. **Duplicate Data Removal:** Remove duplicate rows of data and observations of the same small-body. Multiple observations of the same small-body skew the sampling data in regression and create a data leakage problem with artificial accuracy in the case where an small-body observation ends up in the training and testing datasets.
1. **Target Variable Isolation (diameter):** All entries without a `diameter` have been removed. `Diameter` (`Y`) is the `target` variable and is required for the training model and to calculate the coefficient of determination *(R<sup>2</sup><sub>test</sub>)* during out-of-sample testing.
1. **Exclusion of High-Cost Features (albedo):**  `Albedo` represents a Group 3 characteristic and was explicitly dropped from the dataset. Removing this feature ensures the model does not rely on cost-prohibited data and maintains operational cost-effective analysis. This also prevents data leakage between `albedo` and the `target` `diameter`.
1. **Feature Subset Selection:** The model requires the `target` `diameter` and only features that fall into Group 1 *(e.g., `semi-major axis` `a`, `eccentricity` `e`, `inclination` `i`)* and Group 2 *(`absolute magnitude` `H`)*.
1. **Feature Completeness:** Entries with missing values in the selected feature subsets were removed. This ensures the regression algorithms receive a complete feature matrix and that any observed variance in R<sup>2</sup><sub>test</sub> is attributable to specific feature subsets and not fluctuations in sample size or missing data.
1. **Target Variable Distribution Scaling:** Scale diameter *(Y)* using a base-10 logarithmic formula. Small-bodies size distribution is non-linear. Scaling the `diameter` (`Y`) ensures that the model weights small and large small-bodies the same.
1. **Outlier Detection:** Because outliers represent a real small-body and not an entry error or out-of-scope feature, outliers will be reviewed but may not be removed just on the nature of being an outlier.
1. **Feature Independence:** Remove any highly correlated  features to reduce the risk of improper feature ranking between the correlated features.

#### 9. Outlier Detection
Calculated the lower and upper bounds from the `interquartile range` (`IQR`) and reviewed all outliers. No entries were determiend to be errors so outliers were kept in the dataset because they represent a real small-body that is part of the population of data. The `diameter` was reviewed to compare to `diameter_log` but will not be used in the model.

```python
q1 = df_clean.quantile(0.25)
q3 = df_clean.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

total_rows = len(df_clean)

for col in df_clean.select_dtypes(include=['number']).columns:
    low_count = (df_clean[col] < lower_bound[col]).sum()
    high_count = (df_clean[col] > upper_bound[col]).sum()

    low_pct = (low_count / total_rows) * 100
    high_pct = (high_count / total_rows) * 100
    
    print(f"{col} -> Lower Outliers: {low_count} ({low_pct:.2f}%) | Upper Outliers: {high_count} ({high_pct:.2f}%)")
```

### IV. Report

#### 1. Exploratory Data Analysis (EDA)
Preliminary analysis of the dataset shows that the key indicators of a small-body size (`target` `diameter` (`Y`)) are `absolute magnitude` (`H`), `semi-major axis` (`a`), `eccentricity` (`e`), and `inclination` (`i`). There is a strong `non-linear` relationship between `diameter` and `H` indicating that `H` is a strong predictor of `diameter`. While there were no `linear` correlations between individual `a`, `e`, `i` features and `diameter` or `H`, comparison of `a`, `e`, and `i` mapped to `diameter` and  `a`, `e`, and `i` mapped to `H` provided similar results. This provides strong evidence that  `a`, `e`, and `i` could meet the operational threshold to predict `diameter` without the need for `H`. The `EDA` alone is not enough to conclude if the `null hypothesis` or the `alternative hypothesis` are correct, but there is a pattern and relationship between `a`, `e`, `i`, and `diameter`.

#### 2. Feature Determination
Analysis of the features indicates that `a`, `e`, and `i` are strong candidates for the regression model because they have a `non-linear`, `structural` relationship to the target. The feature `q` is reserved for future determination for inclusion or exclusion in the regression model since it is derived from `a` and `e` which could introduce `multicollinearity` or `noise`. Features `om` and `w` can be excluded. While `om` and `w` are not derived from any other features, the lack of correlation to the `target` may introduce `noise` to the regression model that could lead to `overfitting`.

#### 3. Model Selection
An analysis of the dataset and feature determination support using a `random forest` regression model to test the hypothesis. Small-bodies form dense structural clustering in the feature space and a highly skewed `target` distribution. Small-bodies also can be observed at great distances which creates mathematical `outliers`. Physical characteristics and derived characteristics may have no correlation to the size of a small-body or introduce `multicollinearity`. A `random forest` regression model uses piecewise step-function approximations to handle extreme `outliers` and `noise` of derived or irrelevant features. This is better than `linear` regression or other regression models that try to fit data into a straight line or curve. This will result in stronger models to increase accuracy to predict small-body sizes and to test the hypothesis.

#### Running Tree-Based Regression Models

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pprint
from asteroid import train_model, predict_model
```

```python
TEST_RUN = True
experiments = [
    ("rf", None, "Random Forest Base"),
    ("xgboost", None, "XGBoost Base"),
    ("keras_wide_deep", {"wide_dim": 1}, "Keras Wide & Deep Neural Network"),
    ("keras_resnet", {"batch_size": 512, "hidden_dim": 256, "num_blocks": 3, "learning_rate": 0.0005}, "Keras Tabular ResNet"),
    ("keras_ft_transformer", {"batch_size": 512, "embedding_dim": 16, "num_heads": 2, "dropout_rate": 0.05, "learning_rate": 0.0003}, "FT-Transformer"),
    ("keras_gated_resnet", {"batch_size": 512, "hidden_dim": 256, "num_blocks": 3, "learning_rate": 0.0005}, "Secure Gated ResNet (Noise-Filtering Architecture)")
]

for i, (engine, overrides, label) in enumerate(experiments, 1):
    mode_label = "Testing" if TEST_RUN else "Training"
    print(f"\n--- [Run {i}/{len(experiments)}] {mode_label} {label} ---")
    active_overrides = overrides.copy() if overrides else {}
    
    if TEST_RUN:
        if engine.startswith("keras_") or engine in ["wide_deep", "resnet", "gated_resnet", "ft_transformer"]:
            active_overrides.update({"epochs": 1, "patience": 1, "batch_size": 64})
        elif engine in ["rf", "xgboost", "xgb", "gradient"]:
            active_overrides.update({"n_estimators": 2, "max_depth": 2})
    train_model(engine_type=engine, override_params=active_overrides, delta_threshold=0.05, test_mode=TEST_RUN)

print(f"\n{len(experiments)} experiment run(s) completed.")
```

#### Model Predictions & Analysis

```python
asteroids = [
    {"name": "Ceres", "actual": 939.4, "data": {"a": 2.766, "e": 0.0797, "i": 10.59, "H": 3.34}},
    {"name": "Pallas", "actual": 513.0, "data": {"a": 2.77, "e": 0.2307, "i": 34.93, "H": 4.12}},
    {"name": "Juno", "actual": 246.596, "data": {"a": 2.671, "e": 0.2557, "i": 12.99, "H": 5.19}}
]

results_records = []
for ast in asteroids:
    name_ast, actual_dia, sample_data = ast["name"], ast["actual"], ast["data"] 
    engines_list = [
        ("rf", "Random Forest"), 
        ("xgboost", "XGBoost"),
        ("keras_wide_deep", "Keras Wide & Deep"),
        ("keras_resnet", "Keras Tabular ResNet"),
        ("keras_ft_transformer", "Keras FT-Transformer"),
        ("keras_gated_resnet", "Keras Gated ResNet")
    ]
    
    for engine, engine_name in engines_list:
        preds = predict_model(sample_data, engine_type=engine, model_type="both", test_mode=TEST_RUN)
        full_p, rest_p = preds['full_model_prediction'], preds['restricted_model_prediction']
        full_err_pct = (abs(full_p - actual_dia) / actual_dia) * 100
        rest_err_pct = (abs(rest_p - actual_dia) / actual_dia) * 100
        results_records.append({
            "Asteroid": name_ast, "Actual Diameter (km)": actual_dia, "Engine": engine_name,
            "Full Pred (km)": round(full_p, 2), "Full Error %": round(full_err_pct, 1),
            "Restricted Pred (km)": round(rest_p, 2), "Restricted Error %": round(rest_err_pct, 1)
        })

df_summary = pd.DataFrame(results_records)
print("\nOVERALL PREDICTION SUMMARY TABLE:\n" + "=" * 95)
print(df_summary.to_string(index=False))
print("=" * 95 + "\n\nMean Absolute Percentage Error (MAPE) by Engine for Full Model:")
print(df_summary.groupby("Engine")["Full Error %"].mean().map("{:.2f}%".format))
```

#### Modeling & Prediction Insights

##### Hypothesis Testing
Acceptance of **Statistical Alternative Hypothesis (H<sub>1,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; 0 <br> The inclusion of Group 2 features yields an increase in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do** increase the accuracy of predicting the `target` `diameter` (`Y`).
* All four models resulted in a &Delta; R<sup>2</sup><sub>test</sub> of no less than `0.4917`. This exceeded the statistical threshold of zero (`0`). This means the expanded features did increase the accuracy of predicting the `target` `diameter` (`Y`).

Acceptance of **Operational Null Hypothesis (H<sub>0,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; &#948; <br> The inclusion of Group 2 features yield an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is not** sufficient enough for operational use.
* All four models resulted in a &Delta; R<sup>2</sup><sub>test</sub> of no less than `0.4917`. This exceeded the operational threshold of `0.05`. This means the restricted feature subset is insufficient for production deployment.

##### Model Performance & Selection
Acceptance of `Gradiant Boosting` (`gbr`) tree-based model over `Random Forest` (`rf`) tree-based model. `GBR` resulted in a `MAPE` of `9.23%` versus `RF` with a `MAPE` of `20.63%`. This means that `GBR` missed predicting the true size of the small-body by `9.23%` where `RF` did `20.63%`.

`GBR` was a better model for this use case because it used sequential error correction versus independent averaging like in `RF`. The ability for `GBR` to learn from prior tree mistakes, especially in the heavily skewed dataset, made it better at building a model to more accurately predict the `target`.

##### Key Insights & Limitations
The training and prediction testing shows that more in-depth and sophisticated training is needed to predict `target` `diameter` with both the full and restrictive subset. While the full feature subset models had R<sup>2</sup><sub>test</sub> values around `0.85`, the `RF` prediction of `Ceres` is a clear indication that there are still issues with the model. This is likely due to the skewed nature of small-bodies where some are very large and the bulk are very small. This skew makes it hard for even tree-based models to accurately predict small-body `diameter` with the given feature set.

This is also evident in the `MAE` of the full feature subset. The values for all models ranged from `1.0341`-`1.0442` while the predictions were significantly off from these averages. This means that the skew of small-body distribution artificially suppresses the global `MAE` downward.
