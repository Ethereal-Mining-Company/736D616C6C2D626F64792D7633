## Problem Statement & Hypothesis Formulation

### I. PROBLEM STATEMENT
<a id="problem-statement"></a>
`Asteroid` `diameter` prediction methods:
* **(1)** using a combination of `very high data collection cost`, `very high predictive power`, and `low observational reliability` features like `albedo` (Group 3) with `low data collection cost`, `medium observational reliability` features, like `absolute magnitude` (`H`) (Group 2)
* **(2)** using a combination of `low data collection cost`, `high observational reliability` features (Group 1) with `low data collection cost`, `medium observational reliability` features, like `absolute magnitude` (`H`) (Group 2)(Ge, 2026; Murcia, 2023; Robinson, 2024).

The project looks to determine if the `diameter` of an `asteroid` can accurately be predicted with only Group 1 features to maintain the `low data collection cost`, `high observational reliability` of features that are inexpensive and not variable or prone to error while observing.

### II. HYPOTHESIS FORMULATION
*Suggested explanation to the problem statement.* \
The project utilizes a two-model comparison framework that will be assessed from a statistical and operational perspective. The regression models `target` `diameter` (`Y`) using two feature subsets: an expanded set of `low-cost`, `high-reliability` observations with `low-cost`, `medium-reliability` observations (**X**<sub>full</sub>) and a restricted set of only `low-cost`, `high reliability` observations (**X**<sub>restricted</sub>). The restricted model (**X**<sub>restricted</sub>) is preferred due to the lower number and higher realiability of features. Current peer-reviewed research dictates that `absolute magnitude` (`H`), a Group 2 feature, is mathematically required to calculate `diameter` (Ge, 2026). 

The hypothesis determines if small-body `diameter` (`Y`) can be accurately predicted within a statistical and operational threshold ($\delta$) by the restrictive feature subset compared to the full feature subset, quantified as:
<div align="center">
  &Delta; R<sup>2</sup><sub>test</sub> = R<sup>2</sup><sub>Model(<b>X</b><sub>full</sub>)</sub> - R<sup>2</sup><sub>Model(<b>X</b><sub>restricted</sub>)</sub>
</div>
<br>

**Statistical Hypothesis** \
The statistical hypothesis tests if the restrictive feature subset of Group 1 features maintains a statistically acceptable degree of accuracy in predicting `target` `diameter` (`Y`) relative to the expanded feature subset of Group 1 and Group 2 features.
* **Statistical Null Hypothesis (H<sub>0,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &le; 0 <br> The inclusion of Group 2 features yield no improvement or a decrease in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do not** increase the accuracy of predicting the `target` `diameter` (`Y`).
* **Statistical Alternative Hypothesis (H<sub>1,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; 0 <br> The inclusion of Group 2 features yields an increase in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do** increase the accuracy of predicting the `target` `diameter` (`Y`).

**Operational Hypothesis** \
The operational hypothesis tests if the restrictive feature subset of Group 1 features maintains an operationally and practically acceptable degree of accuracy ($\delta$) in predicting `target` `diameter` (`Y`) relative to the expanded feature subset of Group 1 and Group 2 features.
* **Operational Null Hypothesis (H<sub>0,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; &#948; <br> The inclusion of Group 2 features yield an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is not** sufficient enough for operational use.
* **Operational Alternative Hypothesis (H<sub>1,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &le; &#948;<br> The inclusion of Group 2 features does not yields an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is** sufficient enough for operational use.

### II. PREPARE
<a id="prepare"></a>
Data has been prepared for regression analysis in the following ways:
1. **Data Structure Integrity:** Validate all data types are correct. Name columns should be `strings` (`str`), continuous values should be `floats` (`float64`), and discrete values should be `integers` (`int`).
1. **Data Domain Validation:** Remove data is implausible such as negative `diameter`.
1. **Duplicate Data Removal:** Remove duplicate rows of data and observations of the same small-body. Multiple observations of the same small-body skew the sampling data in regression and create a data leakage problem with artificial accuracy in the case where an small-body observation ends up in the training and testing datasets.
1. **Target Variable Isolation (diameter):** All entries without a `diameter` have been removed. `Diameter` (`Y`) is the `target` variable and is required for the training model and to calculate the coefficient of determination *(R<sup>2</sup><sub>test</sub>)* during `out-of-sample` testing.
1. **Exclusion of High-Cost Features (albedo):**  `Albedo` represents a Group 3 characteristic and was explicitly dropped from the dataset. Removing this feature ensures the model does not rely on cost-prohibited data and maintains operational cost-effective analysis.
1. **Feature Subset Selection:** The model requires the `target` `diameter` and only features that fall into Group 1 *(e.g., `semi-major axis` `a`, `eccentricity` `e`, `inclination` `i`)* and Group 2 *(`absolute magnitude` `H`)*.
1. **Feature Completeness:** Entries with missing values in the selected feature subsets were removed. This ensures the regression algorithms receive a complete feature matrix and that any observed variance in R<sup>2</sup><sub>test</sub> is attributable to specific feature subsets and not fluctuations in sample size or missing data.
1. **Target Variable Distribution Scaling:** Scale `diameter` (`Y`) using a `base-10` `logarithmic` formula. Small-bodies size distribution is non-linear. Scaling the `diameter` (`Y`) ensures that the model weights small and large small-bodies the same.
1. **Outlier Detection:** Because outliers represent a real small-body and not an entry error or out-of-scope feature, outliers will be reviewed but may not be removed just on the nature of being an outlier.
1. **Feature Independence:** Remove any highly correlated  features to reduce the risk of improper feature ranking between the correlated features.

#### 9. Outlier Detection
Calculated the lower and upper bounds from the `interquartile range` (`IQR`) and reviewed all outliers. No entries were determiend to be errors so outliers were kept in the dataset because they represent a real small-body that is part of the population of data. The `diameter` was reviewed to compare to `diameter_log` but will not be used in the model.

```python
q1 = df_clean.quantile(0.25)
q3 = df_clean.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

total_rows = len(df_clean)

for col in df_clean.select_dtypes(include=['number']).columns:
    low_count = (df_clean[col] < lower_bound[col]).sum()
    high_count = (df_clean[col] > upper_bound[col]).sum()

    low_pct = (low_count / total_rows) * 100
    high_pct = (high_count / total_rows) * 100
    
    print(f"{col} -> Lower Outliers: {low_count} ({low_pct:.2f}%) | Upper Outliers: {high_count} ({high_pct:.2f}%)")
```

#### 1. Exploratory Data Analysis (EDA)
Preliminary analysis of the dataset shows that the key indicators of a small-body size (`target` `diameter` (`Y`)) are `absolute magnitude` (`H`), `semi-major axis` (`a`), `eccentricity` (`e`), and `inclination` (`i`). There is a strong `non-linear` relationship between `diameter` and `H` indicating that `H` is a strong predictor of `diameter`. While there were no `linear` correlations between individual `a`, `e`, `i` features and `diameter` or `H`, comparison of `a`, `e`, and `i` mapped to `diameter` and  `a`, `e`, and `i` mapped to `H` provided similar results. This provides strong evidence that  `a`, `e`, and `i` could meet the operational threshold to predict `diameter` without the need for `H`. The `EDA` alone is not enough to conclude if the `null hypothesis` or the `alternative hypothesis` are correct, but there is a pattern and relationship between `a`, `e`, `i`, and `diameter`.

#### 2. Feature Determination
Analysis of the features indicates that `a`, `e`, and `i` are strong candidates for the regression model because they have a `non-linear`, `structural` relationship to the target. The feature `q` is reserved for future determination for inclusion or exclusion in the regression model since it is derived from `a` and `e` which could introduce `multicollinearity` or `noise`. Features `om` and `w` can be excluded. While `om` and `w` are not derived from any other features, the lack of correlation to the `target` may introduce `noise` to the regression model that could lead to `overfitting`.

#### 3. Model Selection
An analysis of the dataset and feature determination support using a `random forest` regression model to test the hypothesis. Small-bodies form dense structural clustering in the feature space and a highly skewed `target` distribution. Small-bodies also can be observed at great distances which creates mathematical `outliers`. Physical characteristics and derived characteristics may have no correlation to the size of a small-body or introduce `multicollinearity`. A `random forest` regression model uses piecewise step-function approximations to handle extreme `outliers` and `noise` of derived or irrelevant features. This is better than `linear` regression or other regression models that try to fit data into a straight line or curve. This will result in stronger models to increase accuracy to predict small-body sizes and to test the hypothesis.

### III. ANALYZE (MODELS)

Ran six regression models on the cleaned dataset. These models were chosen because they can best handle the non-linear, heavily skewed data of asteroids. The two classical machine learning models are `Random Forest` (`RF`) and `Gradient Boosting Regression` (`GBR`, `XGBoost` in code) which utilize decision-trees to ask questions to split data into distinct numerical groups to calculate a continuous prediction value. 

The four deep learning models are `Wide & Deep Network` (`W&D`), `Tabular ResNet` (`TabResNet`), `FT-Transformer` and `Gated ResNet` (`GRN`) used because they handle tabular data like in the structured dataset. `W&D` is used because the wide path can learn the known mathematical rules like the correlation of `absolute magnitude` (`H`) and `target` `diameter` while the deep path can learn any subtle patterns with the orbital characteristics (`a`, `e`, `i`). `TabResNet` minimizes the effects of vanishing gradiant problem as the network gets deeper. `FT-Transformer` was selected to map the complex dependencies of the features together. It looks at them as a unified system rather than isolated variables. `GRN` was selected because it is similar to `TabResNet` but will use gates to filter out noisy data to help predict the patterns in orbital characteristics, specifically in the restricted subset.

Each model was trained with the full and restricted subsets and the results logged for analysis. Training utilized a five-fold cross validation where each fold and fold results were logged as well.

#### Model Training

The models were trained and tested against the hypotheses, quantified as:
<div align="center">
  &Delta; R<sup>2</sup><sub>test</sub> = R<sup>2</sup><sub>Model(<b>X</b><sub>full</sub>)</sub> - R<sup>2</sup><sub>Model(<b>X</b><sub>restricted</sub>)</sub>
</div>
<br>

&Delta; R<sup>2</sup><sub>test</sub> was used to reject or fail to reject the null hypotheses. Every model resulted in &Delta; R<sup>2</sup><sub>test</sub> greater than the statistical and operational threshold ($\delta$ = 0.05). This means that the restricted features subset was not sufficient to accurately predict the `diameter` of an `asteroid` for operational use. As such, final model determination was based on the greatest full feature subset model mean R<sup>2</sup>. While all models met or exceeded a mean R<sup>2</sup> of `0.7887`, it was the traditional models of `Random Forest` and `Gradient Boosted Regression` that scored the best with a mean R<sup>2</sup> of `0.8508` and `0.8548` respectively. The determination of which model would be used in production will undergo an analysis of `Mean Absolute Percentage Error` (`MAPE`) to determine how well predictions matched actual values of `asteroids`. The model with the lowest `MAPE` will be selected for use in production.

```python
experiments = [
    ("rf", None, "Random Forest Base"),
    ("xgboost", None, "XGBoost Base"),
    ("keras_wide_deep", {"wide_dim": 1}, "Keras Wide & Deep Neural Network"),
    ("keras_resnet", {"batch_size": 512, "hidden_dim": 256, "num_blocks": 3, "learning_rate": 0.0005}, "Keras Tabular ResNet"),
    ("keras_ft_transformer", {"batch_size": 512, "embedding_dim": 16, "num_heads": 2, "dropout_rate": 0.05, "learning_rate": 0.0003}, "FT-Transformer"),
    ("keras_gated_resnet", {"batch_size": 512, "hidden_dim": 256, "num_blocks": 3, "learning_rate": 0.0005}, "Gated ResNet (Noise-Filtering Architecture)")
]

for i, (engine, overrides, label) in enumerate(experiments, 1):
    mode_label = "Testing" if TEST_RUN else "Training"
    print(f"\n--- [Run {i}/{len(experiments)}] {mode_label} {label} ---")
    active_overrides = overrides.copy() if overrides else {}
    
    if TEST_RUN:
        if engine.startswith("keras_") or engine in ["wide_deep", "resnet", "gated_resnet", "ft_transformer"]:
            active_overrides.update({"epochs": 1, "patience": 1, "batch_size": 64})
        elif engine in ["rf", "xgboost", "xgb", "gradient"]:
            active_overrides.update({"n_estimators": 2, "max_depth": 2})
    train_model(engine_type=engine, override_params=active_overrides, delta_threshold=0.05, test_mode=TEST_RUN)

print(f"\n{len(experiments)} experiment run(s) completed.")
```

```python
df_logs = pd.read_csv('logs/prediction_results.csv')

conditions = [
    (df_logs['Actual Diameter (km)'] < 1.0),
    (df_logs['Actual Diameter (km)'].between(1.0, 50.0)),
    (df_logs['Actual Diameter (km)'] > 50.0)
]
choices = ['Sub-Kilometer (<1km)', 'Medium (1-50km)', 'Massive (>50km)']
df_logs['Size_Bucket'] = np.select(conditions, choices, default='Unknown')
df_logs = df_logs.sort_index(kind='stable')
if 'Engine' in df_logs.columns and 'Subset' in df_logs.columns:
    df_logs = df_logs.sort_values(by=['Engine', 'Subset', 'Size_Bucket'], kind='stable').reset_index(drop=True)

column_order = [
    'Full: Avg MAPE', 'Full: Error Volatility (Std)',
    'Restricted: Avg MAPE', 'Restricted: Error Volatility (Std)'
]

matrix_blocks = []
unique_engines = sorted(df_logs['Engine'].unique())

for eng in unique_engines:
    df_eng = df_logs[df_logs['Engine'] == eng]
    
    df_variance = df_eng.groupby(['Engine', 'Subset', 'Size_Bucket'])['Error %'].agg(
        Avg_MAPE='mean', 
        Std_Volatility='std'
    )
    
    df_pivoted = df_variance.unstack(level='Subset')
    df_pivoted.columns = [
        'Full: Avg MAPE', 'Restricted: Avg MAPE',
        'Full: Error Volatility (Std)', 'Restricted: Error Volatility (Std)'
    ]
    df_pivoted = df_pivoted[column_order]
    matrix_blocks.append(df_pivoted)
    
    df_sub_raw = df_eng.groupby(['Subset'])['Error %'].agg(
        Avg_MAPE='mean', 
        Std_Volatility='std'
    )
    
    sub_data = {
        'Full: Avg MAPE': [df_sub_raw.loc['full', 'Avg_MAPE'] if 'full' in df_sub_raw.index else np.nan],
        'Full: Error Volatility (Std)': [df_sub_raw.loc['full', 'Std_Volatility'] if 'full' in df_sub_raw.index else np.nan],
        'Restricted: Avg MAPE': [df_sub_raw.loc['restricted', 'Avg_MAPE'] if 'restricted' in df_sub_raw.index else np.nan],
        'Restricted: Error Volatility (Std)': [df_sub_raw.loc['restricted', 'Std_Volatility'] if 'restricted' in df_sub_raw.index else np.nan]
    }
    
    df_sub_avg = pd.DataFrame(sub_data, index=pd.MultiIndex.from_tuples([(eng, 'Average')]))
    df_sub_avg = df_sub_avg[column_order]
    matrix_blocks.append(df_sub_avg)

df_final_matrix = pd.concat(matrix_blocks)

print("\n" + "="*95 + "\nMAPE & VOLATILITY MATRIX WITH MODEL\n" + "="*95)
display(df_final_matrix.style.format("{:.2f}%", na_rep="N/A"))
```

#### 1. Hypothesis Testing
**Acceptance of Statistical Alternative Hypothesis (H<sub>1,stat</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; 0 <br> 
* *Definition:* The inclusion of Group 2 features yields an increase in the out-of-sample &Delta; R<sup>2</sup><sub>test</sub>, indicating that the expanded features **do** increase the accuracy of predicting the `target` `diameter` (`Y`).
* *Result:* All six models resulted in a &Delta; R<sup>2</sup><sub>test</sub> of no less than `0.4917`. This exceeded the statistical threshold of zero (`0`). This means the expanded features did increase the accuracy of predicting the `target` `diameter` (`Y`).

**Acceptance of Operational Null Hypothesis (H<sub>0,ops</sub>):** &Delta; R<sup>2</sup><sub>test</sub> &gt; &#948; <br> 
* *Definition:* The inclusion of Group 2 features yield an operationally significant increase in accuracy to predict `target` `diameter` (`Y`), indicating that the restricted feature subset **is not** sufficient enough for operational use.
* *Result:* All six models resulted in a &Delta; R<sup>2</sup><sub>test</sub> of no less than `0.4917`. This exceeded the operational threshold of `0.05`. This means the restricted feature subset is insufficient for production deployment.

#### 2. Feature Variance Analysis
* Feature importance and sensitivity metrics confirm that `absolute magnitude` (`H`) accounts for `76.74%` of the variance explained when predicting `diameter`. Omitting `H` from the training pipeline more than triples the error rate across all models, demonstrating that the geometric elements (`a`, `e`, `i`) lack sufficient standalone signal and `H` is required to reliably and accurately predict `diameter`.

#### 3. Model Performance & Selection
* **Final Model Selection:** `Random Forest` (`RF`) 
    * `RF` is formally selected as the most reliable deployment model for predicting `asteroid` `diameter`. Across five-fold cross-validation on the full feature set, `RF` achieved a Mean R² of `0.8508` and an `out-of-sample` Mean `MAE` of `1.0443 km`. While `Gradient Boosting Regression` (`GBR`, `XGBOOST` in code) demonstrated a slightly higher Mean R² of `0.8548` and a comparable Mean `MAE` of `1.0360 km`, `RF` was ultimately selected for production deployment because it resulted in the lowest overall average `Mean Absolute Percentage Error` (`MAPE`) of `14.44%` compared to the average `MAPE` for `GBR` of `20.80%`.

#### 4. Classical Modeling Insights
* Classical tree-based machine learning models performed the best for highly skewed asteroid data. This is likely because it utilizes discrete, step-based data partitioning to isolate numeric subgroups for predicting `diameter`. Tree-based models are limited because they can only predict within the range of their training data but were able to establish more reliable patterns in that range for predicting `diameter`. For instance, while `GBR` handled extreme values better at the `Massive` (>50km) scale (`8.10%` vs `18.03%` `MAPE`), `RF` outperformed it at the highly populous `Sub-Kilometer` (<1km) scale (`20.37%` vs `35.98%` `MAPE`), securing the lowest global error average.

#### 5. Neural Network Insights
* Neural networks struggled to reliably predict `diameter` because they favor optimizing for generalized pattern fitting. The highly skewed dataset and inconsistent measurements of features like `H` were likely over-smoothed. This caused the models to overpredict at the `Sub-Kilometer` scale and struggle to handle the `Massive` scale. Furthermore, removing `H` caused complete mathematical failures during training. Without this feature, both the `W&D` and `GRN` models completely broke down on certain `asteroid` scales, dropping to negative accuracy levels and throwing massive errors exceeding 10<sup>21</sup> kilometers.
